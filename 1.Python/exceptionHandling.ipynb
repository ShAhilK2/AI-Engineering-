```python
# try and except 

# try => try is used to write a risky code and except handles the error if something went wrong

try:
    num = int(input("Enter a number: "))
    print(num)
except ValueError:
    print("Invalid input")
else:
    print("iF TRY IS SUCCESSFULLY RUN",num)
finally:
    print("This is finally block")
```

```python
try:
    div = 10/0
except ValueError:
    print("Invalid input")
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("run if success")
finally:
    print("This is finally block")

```

```python
# raise 

def withdraw_money(balance,amount):
    if amount > balance:
        raise ValueError("Insufficient funds")
    return balance - amount

withdraw_money(100,200)
```

```python
# Custom Exception helps us to create our own exception/erros

class InsufficientFundsError(Exception):
    print("Please Managa your minimum balance")
 
def withdraw_money(balance, amount):
    if amount > balance:
        raise InsufficientFundsError("Not Enough balance")
    return balance - amount


try:
    withdraw_money(100, 200)
except InsufficientFundsError as e:
    print("Error:", e)
```

```python
# Exception Chaining is used when one exception happens because of another exception

def  convert_into_int(values):
    try:
        return int(values)
    except ValueError as e:
        raise ValueError("Value must be integer") from e


convert_into_int("state")
```
