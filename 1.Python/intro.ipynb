Python Identifier 
Identifiers are the names given to variables,classes and methods

Keywords 
Keywords are unique predefined,reserved words

<!-- Rules for Naming Identifier -->

1. Identifiers are case sensitive 
2. It can have squences of letter and digits.However it must always begin with a letter or an underscore(_)
3. It cannot start with a number/digit.
4. It cannot contain special characters like @, #, $, %, etc.
5. It cannot be a keyword
6. Whitespaces are not allowed


Make sensible names as they make your code more readable and maintainable

```python


# Set 
# Set is an unordered collection of unique elements and speared by comma enclosed in {},and cannot get each by indexing 
unique_numbers = {1,2,3,4,5}

unique_numbers.add(6)
unique_numbers.remove(1)
unique_numbers.update([7,8,9])
unique_numbers.discard(2)
print(unique_numbers)
print(len(unique_numbers))
print(type(unique_numbers))
```
