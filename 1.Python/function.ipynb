```python
# Function => Function are reusbale block of code to perform a specific task 
# they help reduce repetition and code modular 


def greet():
    print("Hello")
greet()
```

```python
# Positional arguments are passed  based on the order of parameters in the function 

def introduce(name, age):
    print(f"Helle {name}, age {age} year old")

introduce("Shahil", 25)

# Keyword Argument  passed  using paramter names,so the order does not matter
introduce(age=23, name="Shahil")
```

```python
# default argument are used when no paramter is passed

def greet(name="Shahil"):
    print(f"Hello {name}")

greet()
greet("Hetvi")
```

```python
# args => args allows function to accept any number of positional arguments .It store in tuples

def add_numbers(*args):
    return sum(args)

print(add_numbers(1, 2, 3, 4, 5))
print(add_numbers(10, 20, 30))
```

```python
# kwargs => kwargs allows function to accept any number of keyword arguments.It store in dictionary

def show_profile(**kwargs):
    print(kwargs)

show_profile(name="Shahil", age=25, city="Ahmedabad")

```

```python
# return keywords sends a value back from a function to the place where it is called

def square(num):
    return num * num

result = square(5)
print(result)
```

```python
def operations(a,b):
    return a+b, a-b, a*b, a/b   


add,sub,mul,div = operations(10,5)
print(add)
print(sub)
print(mul)
print(div)
```

```python
# SCOPE
# Local variable created inside a function belongs to the local scope of that function, and can only be used inside that function.

def myfunc():
     local_var = "I am local"
     print(local_var)


myfunc()

#Global Variable :
# A variable created in the main body of the Python code is a global variable and belongs to the global scope.
# Global variables are available from within any scope, global and local.

global_var = "I am global"

def myfunc():
    print(global_var)

myfunc()


count = 0 

def increment():
    global count
    count += 1
    return count

print(increment())

# non local => nested function 
def outer():
    x = 10
    def inner():
        nonlocal x
        x += 1
        return x
    return inner()

print(outer())




```

```python
# LAMBDA EXPRESSION : => are small anonymous function used for a short one-line operations

square = lambda x: x * x
is_even = lambda x: x % 2 == 0 
to_upper = lambda s: s.upper()   

print(square(5))
print(is_even(11))
print(to_upper("hello"))
```

```python
# DocString => are used to describe what functions does 

def add_numbers(a,b):
    """Add number and return it"""
    return a+b

print(add_numbers(10,20))
```

```python
# Type Annotation => describe the expected the input and output type of a function

def add_numbers(a: int, b: int) -> int:
    return a + b
print(add_numbers(5, 10))
```
